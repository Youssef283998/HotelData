```python
df['booking_status']=df['booking_status'].map({'Not_Canceled':'1',
                                               'Canceled':'0'})
```

```python
df.columns
```

```python
features = [ 
    'lead_time',
    'avg_price_per_room',
    'repeated_guest',
    'no_of_special_requests',
    'total_guest',
    'total_nights',

]
```

```python
x=df[features]
```

```python
y=df['booking_status']
```

```python


from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score, precision_score, recall_score
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_absolute_percentage_error
from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import RandomizedSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

from sklearn.utils.class_weight import compute_class_weight
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0,stratify=y)

```

```python
# 1. تحويل y_train إلى int
y_train = y_train.astype(int)
```

```python
classes =np.unique(y_train)
```

```python
classes
```

```python
weights= compute_class_weight( class_weight='balanced', classes=classes , y= y_train)
```

```python
weights
```

```python
class_weights= dict(zip(classes, weights))
```

```python
class_weights
```

```python
rf= RandomForestClassifier(class_weight= class_weights)
```

```python
rf_params= {
    "n_estimators":[ 100,200,300],
    "max_depth":[5,10,20,30],
    "min_samples_split":[2,5,10,14],
    "min_samples_leaf":[1,2,3,4],
    
}
```

```python
rf_random= RandomizedSearchCV(
    rf,
    rf_params,
    n_iter=20,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
)
```

```python
rf_random.fit(x_train, y_train)
```

```python
rf_random.best_params_
```

```python
gb= GradientBoostingClassifier()
```

```python
gb_params ={
    "n_estimators": [100,200,500],
    "max_depth": [3,5,10,25],
    "learning_rate": [0.1,0.05,0.01,0.001,0.002,0.02]
   
}
```

```python
gb_random = RandomizedSearchCV(
    gb,
    gb_params,
    n_iter=20,
    cv=3,
    scoring='accuracy',
    n_jobs=-1
)
```

```python
gb_random.fit(x_train, y_train)
```

```python
xgp=XGBClassifier(
    scale_pos_weight= class_weights [0] / class_weights[1],
    eval_matric= 'logloss'


)
```

```python
xgp_params= {
    "n_estimators": [100,300,700],
    "max_depth":[3,5,7,15],
    "learning_rate" :[0.1,0.2,0.01,0.02,0.001],
    "subsample" : [0.6,0.1],
    "colsample_bytree" : [0.6,0.1],

}
```

```python
xgp_random=RandomizedSearchCV(
    xgp ,
    xgp_params ,
    n_iter =16 ,
    cv = 5 ,
    scoring = 'accuracy' ,
    n_jobs = -1 ,
)
```

```python
xgp_random.fit(x_train, y_train)
```

```python
models={
    'RandomForest': rf_random.best_estimator_,
    'GradientBoosting' : gb_random.best_estimator_,
    'XGBoost' : xgp_random.best_estimator_,
    }
```

```python
for name, model in models.items():
    print(name)
    y_pred = model.predict(x_test)
    
    # تحويل y_test إلى أرقام صحيحة لتتطابق مع y_pred
    y_test_converted = y_test.astype(int)
    
    print(accuracy_score(y_test_converted, y_pred))
```
